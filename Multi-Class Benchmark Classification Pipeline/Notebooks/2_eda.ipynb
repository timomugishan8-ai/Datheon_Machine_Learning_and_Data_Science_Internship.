```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flower_df = pd.read_csv("../data/processed/iris_cleaned.csv")
```

# Exploratory Data Analysis

```python
###1. Pairplot of the flower dataset

sns.pairplot(flower_df, hue="species")
plt.show()
```

INTERPRETATION 

-> When you critically observe the diagonal charts, we are analyzing the distribution of a univariate variable. Because most of the groups overlap, we cannot use a single variable as a strong predictor for the target variable, since it cannot tell the 3 groups apart.

-> The groups Versicolor and Virginica overlap more, especially in sepal measurements, but are still reasonably separated in petal dimensions.

-> The off-diagonal scatter plots tell a different story. Even though most of them do separate the 3 groups to a certain degree, the bivariate variables petal_length and petal_length show both the distinction and the direction of the relationship, which gives us a clear straight shape. In short, the scatter plots show strong positive relationships, especially between petal_length and petal_width.

-> The clear species clusters suggest that tree-based models, KNN, or logistic regression should work well. Petal measurements should receive priority during feature analysis.

```python
###2. Correlation heatmap of the flower dataset

plt.figure(figsize=(10, 8))
sns.heatmap(flower_df[["petal_length", "petal_width", "sepal_length", "sepal_width"]].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Flower Dataset")
plt.show()
```

INTERPRETATION

-> The correlation values indicate that all variables strongly and positively correlate with each other apart from sepal_width, which produces values that indicate weak or negative association with the others.

```python
###3. Bar graph showing the distribution of species in the flower dataset
plt.figure(figsize=(8, 6))
sns.countplot(x="species", data=flower_df, palette="Set2")
plt.title("Distribution of Species in Flower Dataset")
plt.show()
```

INTERPRETATION

Setosa and versicolor each contain exactly 50 observations, with virginica containing 49. The dataset is fairly balanced, so accuracy is a reasonable initial metric.
