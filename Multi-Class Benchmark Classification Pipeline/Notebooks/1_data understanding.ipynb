INTERPRETATION 

The iris dataset is one that is favorable for analysis and modeling as a beginner. The size of this dataset is quite small, occupying just 150 rows and 5 columns. This means that we have 150 counts of each variable. The variables include sepal_length, sepal_width, petal_length, petal_width, and the soon-to-be target variable, species.

The first 4 variables that indicate the size of the flower parts are continuous while the species variable is categorical, with 3 classes each having a count of 50 observations. The description table gives us a summary of the descriptive statistics of the dataset.

The dataset is mostly quite clean. It had no missing values, but it did have 1 duplicated row. Because it's only one, I discarded it to ensure the dataset is clean. Having the rechecked, the dataset is now clean. This means now that right now, one of the variables is short by one observation, which is 'virginica'. Regardless, the target is still balanced across the 3 species.

```python
#Save the cleaned dataset to a new CSV file

from pathlib import Path

# Create the processed-data directory if it does not exist
processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

# Save the cleaned dataset
flower_df.to_csv(processed_dir / "iris_cleaned.csv", index=False)
```
